## Big Data Processing & Spark Pipeline


```python
import pyspark
import os

print("PySpark version:", pyspark.__version__)
print("HADOOP_HOME:", os.environ.get("HADOOP_HOME"))
print("winutils exists:", os.path.exists(r"C:\hadoop\bin\winutils.exe"))
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder \
    .appName("Project10_Member2_Pipeline") \
    .config("spark.sql.warehouse.dir", "file:///C:/temp") \
    .getOrCreate()

print("Spark Session ready!")
```

## Data Ingestion

```python
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("../data/sample.csv")

print("Original schema:")
df.printSchema()

print("Total rows before cleaning:", df.count())
```

## Data Transformation & Cleaning

```python
processed_spark_df = (
    df.dropna()
)

print("Total rows after cleaning:", processed_spark_df.count())

processed_spark_df.show(5)

# Save using Spark
processed_spark_df.write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("../spark/final_processed_data")

print("Pipeline completed successfully!")
```
